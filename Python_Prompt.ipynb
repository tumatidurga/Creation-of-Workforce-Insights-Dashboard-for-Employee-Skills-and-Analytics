## 1. Variables and Data Types

```python
# Declaring variables with different data types
student_id = 101 # Integer
student_name = "Alice Smith" # String
student_age = 20 # Integer
student_gpa = 3.9 # Float
is_enrolled = True # Boolean
courses = ["Math", "Science", "History"] # List

# Printing variables and their types
print(f"Student ID: {student_id}, Type: {type(student_id)}")
print(f"Student Name: {student_name}, Type: {type(student_name)}")
print(f"Student Age: {student_age}, Type: {type(student_age)}")
print(f"Student GPA: {student_gpa}, Type: {type(student_gpa)}")
print(f"Is Enrolled: {is_enrolled}, Type: {type(is_enrolled)}")
print(f"Courses: {courses}, Type: {type(courses)}")
```

## 3. Operators

```python
# Arithmetic Operators
credits_completed = 60
credits_needed = 120
credits_remaining = credits_needed - credits_completed
print(f"Credits remaining: {credits_remaining}")
tuition_per_credit = 250.50
total_tuition_due = credits_remaining * tuition_per_credit
print(f"Total tuition due: ${total_tuition_due}")

# Comparison Operators
has_good_gpa = student_gpa > 3.5
is_senior = credits_completed >= 90
print(f"Has good GPA: {has_good_gpa}")
print(f"Is Senior: {is_senior}")

# Logical Operators
is_eligible_for_scholarship = has_good_gpa and not is_senior # Assuming scholarship is for non-seniors with good GPA
print(f"Is eligible for scholarship: {is_eligible_for_scholarship}")
```

## 4. If-Else Statements

## 6. While Loop

```python
# Simulating credits accumulation towards graduation
current_credits = 0
required_credits = 120
print("Accumulating credits...")
while current_credits < required_credits:
    current_credits += 15 # Assuming 15 credits per semester
    print(f"Current credits: {current_credits}")
    if current_credits >= required_credits:
        break # Exit loop if credits meet requirement
print("Graduation requirements met!")
```

## 10. Dictionaries

```python
# Creating a dictionary to store student profiles
student_profiles = {
    101: {"name": "Alice Smith", "age": 20, "gpa": 3.9, "major": "CS"},
    102: {"name": "Bob Johnson", "age": 22, "gpa": 3.5, "major": "Physics"},
    103: {"name": "Charlie Brown", "age": 19, "gpa": 3.2, "major": "Math"}
}
print(f"Initial Student Profiles: {student_profiles}")

# Accessing values
print(f"Alice's GPA: {student_profiles[101]['gpa']}")

# Adding a new student
student_profiles[104] = {"name": "David Lee", "age": 21, "gpa": 3.7, "major": "Chemistry"}
print(f"After adding David: {student_profiles}")

# Updating a value
student_profiles[101]["gpa"] = 4.0
print(f"Alice's updated GPA: {student_profiles[101]['gpa']}")

# Iterating through keys and values
print("\nAll Student Names:")
for student_id, details in student_profiles.items():
    print(f"- {details['name']} (ID: {student_id})")
```

## 11. Sets

```python
# Creating a set of enrolled courses (sets store unique elements)
enrolled_courses_alice = {"Math", "Physics", "Chemistry"}
enrolled_courses_bob = {"Math", "History", "Physics"}
print(f"Alice's courses: {enrolled_courses_alice}")
print(f"Bob's courses: {enrolled_courses_bob}")

# Adding an element (duplicates are ignored)
enrolled_courses_alice.add("Biology")
enrolled_courses_alice.add("Math") # This will not add 'Math' again
print(f"Alice's courses after adding Biology: {enrolled_courses_alice}")

# Removing an element
enrolled_courses_alice.remove("Chemistry")
print(f"Alice's courses after removing Chemistry: {enrolled_courses_alice}")

# Set operations
common_courses = enrolled_courses_alice.intersection(enrolled_courses_bob)
print(f"Common courses between Alice and Bob: {common_courses}")

all_courses = enrolled_courses_alice.union(enrolled_courses_bob)
print(f"All courses taken by Alice or Bob: {all_courses}")
```

## 12. String Methods

```python
student_full_name = "  john doe   "
student_major = "computer science"

# strip() - remove leading/trailing whitespace
cleaned_name = student_full_name.strip()
print(f"Cleaned Name: '{cleaned_name}'")

# capitalize() - capitalize the first letter
capitalized_major = student_major.capitalize()
print(f"Capitalized Major: {capitalized_major}")

# title() - capitalize the first letter of each word
titled_name = cleaned_name.title()
print(f"Titled Name: {titled_name}")

# upper() and lower()
print(f"Uppercase Name: {cleaned_name.upper()}")
print(f"Lowercase Name: {cleaned_name.lower()}")

# split() - split string into a list
name_parts = cleaned_name.split()
print(f"Name Parts: {name_parts}")

# join() - join list elements into a string
joined_name = "-".join(name_parts)
print(f"Joined Name: {joined_name}")

# replace()
email_suffix = cleaned_name.replace(' ', '.') + '@university.edu'
print(f"Email: {email_suffix}")

# startswith() and endswith()
print(f"Name starts with 'john': {cleaned_name.startswith('john')}")
print(f"Major ends with 'science': {student_major.endswith('science')}")
```

## 14. Exception Handling (try-except)

```python
# Example 1: Handling division by zero
def calculate_average_gpa(total_gpa_points, num_students):
    try:
        average = total_gpa_points / num_students
        print(f"Average GPA: {average:.2f}")
    except ZeroDivisionError:
        print("Error: Cannot calculate average GPA with zero students.")
    except TypeError:
        print("Error: Invalid input types for GPA calculation.")

calculate_average_gpa(10.5, 3)
calculate_average_gpa(0, 0)
calculate_average_gpa("invalid", 2)

# Example 2: Handling invalid input for student ID lookup
def get_student_name_safe(s_id):
    try:
        name = student_profiles[s_id]['name']
        print(f"Student name for ID {s_id}: {name}")
    except KeyError:
        print(f"Error: Student with ID {s_id} not found.")

get_student_name_safe(101)
get_student_name_safe(999)
```

## 16. Classes and Objects, 17. Constructors, 20. Encapsulation

```python
class Student:
    # Constructor: __init__ method is called when a new object is created
    def __init__(self, student_id, name, age, major, gpa):
        # Encapsulation: Using double underscore to make attributes 'private' (by convention)
        self.__student_id = student_id
        self.__name = name
        self.__age = age
        self.__major = major
        self.__gpa = gpa
        self.__credits = 0 # Default value

    # Method to display student information
    def display_info(self):
        return (f"Student ID: {self.__student_id}, Name: {self.__name}, Age: {self.__age}, "
                f"Major: {self.__major}, GPA: {self.__gpa:.2f}, Credits: {self.__credits}")

    # Getter method for name (Encapsulation)
    def get_name(self):
        return self.__name

    # Setter method for GPA (Encapsulation)
    def set_gpa(self, new_gpa):
        if 0.0 <= new_gpa <= 4.0:
            self.__gpa = new_gpa
            print(f"GPA for {self.__name} updated to {self.__gpa:.2f}")
        else:
            print("Invalid GPA. Must be between 0.0 and 4.0.")

    # Method to enroll in a course
    def enroll_course(self, course_credits):
        if course_credits > 0:
            self.__credits += course_credits
            print(f"{self.__name} enrolled for {course_credits} credits. Total credits: {self.__credits}")
        else:
            print("Course credits must be positive.")

# Creating objects (instances) of the Student class
alice = Student(101, "Alice Wonderland", 20, "Computer Science", 3.9)
bob = Student(102, "Bob The Builder", 22, "Engineering", 3.5)

# Accessing attributes and calling methods
print(alice.display_info())
print(bob.display_info())

# Using getter and setter
print(f"\nAlice's name (using getter): {alice.get_name()}")
alice.set_gpa(4.0)
alice.set_gpa(5.0) # This will show an error message
alice.enroll_course(3)
alice.enroll_course(4)
print(alice.display_info())
```

## 18. Inheritance

```python
class GraduateStudent(Student):
    # GraduateStudent inherits from Student
    def __init__(self, student_id, name, age, major, gpa, research_topic):
        # Call the constructor of the parent class (Student)
        super().__init__(student_id, name, age, major, gpa)
        self.__research_topic = research_topic # New attribute for GraduateStudent
        self.__thesis_status = "Pending"

    # Override display_info method (Polymorphism example later)
    def display_info(self):
        parent_info = super().display_info() # Get info from parent
        return (f"{parent_info}, Research Topic: {self.__research_topic}, "
                f"Thesis Status: {self.__thesis_status}")

    def submit_thesis(self):
        self.__thesis_status = "Submitted"
        print(f"{self.get_name()} has submitted their thesis on {self.__research_topic}.")

# Creating an object of the GraduateStudent class
charlie = GraduateStudent(201, "Charlie Puth", 25, "AI", 3.7, "Neural Networks")
print(f"\n{charlie.display_info()}")
charlie.enroll_course(5) # Inherited method
charlie.submit_thesis() # GraduateStudent specific method
print(charlie.display_info())
```

## 19. Polymorphism

```python
# Polymorphism: Objects of different classes can be treated uniformly.
# Here, both Student and GraduateStudent objects can be stored in the same list
# and respond to the `display_info` method, even if their implementation differs.

students_list = [alice, bob, charlie]

print("\n--- Displaying all students' info using a loop ---")
for student in students_list:
    # The appropriate display_info method is called based on the object's actual type
    print(student.display_info())
    print("-" * 30)

# Another example of polymorphism: a function that accepts different student types
def process_student(s):
    print(f"Processing: {s.get_name()}")
    s.enroll_course(3) # Can call common methods
    if isinstance(s, GraduateStudent):
        s.submit_thesis() # Can call specific methods if type is known
    print("Done processing.")
    print("-" * 30)

print("\n--- Processing students with a common function ---")
process_student(alice)
process_student(charlie)
```
