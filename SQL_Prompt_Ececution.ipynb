## Student Management System with SQLite

Create a Student Management System. Requirements: 1. Create two tables: Students and Courses. 2. Insert at least 5 records into each table. 3. Perform the following DDL operations: - CREATE TABLE - ALTER TABLE 4. Perform the following DML operations: - INSERT - UPDATE - DELETE 5. Write SQL queries using: - SELECT - WHERE - ORDER BY - GROUP BY - COUNT - AVG 6. Perform an INNER JOIN between the Students and Courses tables. 7. Add comments to explain each step. 8. Display the output of each query. 9. Ensure the code runs successfully in Google Colab using SQLite.


```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
# This database will be cleared when the Colab runtime restarts
conn = sqlite3.connect(':memory:')
c = conn.cursor()

print("SQLite database connected successfully in memory.")
```

### 1. Create Tables: Students and Courses (and Enrollments for JOIN)

```python
# Create Students table
c.execute('''
    CREATE TABLE Students (
        student_id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        age INTEGER
    )
''')

# Create Courses table
c.execute('''
    CREATE TABLE Courses (
        course_id INTEGER PRIMARY KEY,
        course_name TEXT NOT NULL,
        instructor TEXT
    )
''')

# Create Enrollments table to link Students and Courses (for INNER JOIN requirement)
c.execute('''
    CREATE TABLE Enrollments (
        enrollment_id INTEGER PRIMARY KEY,
        student_id INTEGER,
        course_id INTEGER,
        FOREIGN KEY (student_id) REFERENCES Students(student_id),
        FOREIGN KEY (course_id) REFERENCES Courses(course_id)
    )
''')

conn.commit()
print("Tables Students, Courses, and Enrollments created successfully.")
```

### 2. Insert at least 5 records into each table

```python
# Insert records into Students table
c.execute("INSERT INTO Students (name, age) VALUES ('Alice', 20)")
c.execute("INSERT INTO Students (name, age) VALUES ('Bob', 22)")
c.execute("INSERT INTO Students (name, age) VALUES ('Charlie', 21)")
c.execute("INSERT INTO Students (name, age) VALUES ('Diana', 23)")
c.execute("INSERT INTO Students (name, age) VALUES ('Eve', 20)")

# Insert records into Courses table
c.execute("INSERT INTO Courses (course_name, instructor) VALUES ('Math', 'Dr. Smith')")
c.execute("INSERT INTO Courses (course_name, instructor) VALUES ('Physics', 'Prof. Jones')")
c.execute("INSERT INTO Courses (course_name, instructor) VALUES ('Chemistry', 'Dr. Brown')")
c.execute("INSERT INTO Courses (course_name, instructor) VALUES ('History', 'Prof. Green')")
c.execute("INSERT INTO Courses (course_name, instructor) VALUES ('Literature', 'Dr. White')")

# Insert records into Enrollments table (linking students and courses)
c.execute("INSERT INTO Enrollments (student_id, course_id) VALUES (1, 1)") # Alice in Math
c.execute("INSERT INTO Enrollments (student_id, course_id) VALUES (1, 2)") # Alice in Physics
c.execute("INSERT INTO Enrollments (student_id, course_id) VALUES (2, 1)") # Bob in Math
c.execute("INSERT INTO Enrollments (student_id, course_id) VALUES (3, 3)") # Charlie in Chemistry
c.execute("INSERT INTO Enrollments (student_id, course_id) VALUES (4, 4)") # Diana in History
c.execute("INSERT INTO Enrollments (student_id, course_id) VALUES (5, 5)") # Eve in Literature
c.execute("INSERT INTO Enrollments (student_id, course_id) VALUES (2, 5)") # Bob in Literature

conn.commit()
print("Records inserted successfully into Students, Courses, and Enrollments tables.")
```

### 3. DDL Operations: ALTER TABLE

```python
# Add a new column 'email' to the Students table
c.execute('''
    ALTER TABLE Students
    ADD COLUMN email TEXT
''')
conn.commit()
print("Column 'email' added to Students table.")

# Display the updated schema of Students table (optional check)
students_schema = pd.read_sql_query("PRAGMA table_info(Students)", conn)
display(students_schema)
```

### 4. DML Operations: UPDATE and DELETE

```python
# UPDATE: Update Alice's age and add an email
c.execute("UPDATE Students SET age = 21, email = 'alice@example.com' WHERE student_id = 1")
conn.commit()
print("Student record updated (Alice's age and email).")

# Display updated Students table
print("Students table after UPDATE:")
display(pd.read_sql_query("SELECT * FROM Students", conn))

# DELETE: Delete a course (e.g., Physics, which has course_id 2)
c.execute("DELETE FROM Courses WHERE course_id = 2")
conn.commit()
print("Course 'Physics' deleted from Courses table.")

# Display updated Courses table
print("Courses table after DELETE:")
display(pd.read_sql_query("SELECT * FROM Courses", conn))

# Note: Deleting a course that was enrolled in will leave orphaned entries in Enrollments
# For a robust system, CASCADE DELETE would be used, but for this example, we'll keep it simple.
# Let's also delete related enrollment for course_id 2 to clean up:
c.execute("DELETE FROM Enrollments WHERE course_id = 2")
conn.commit()
print("Related enrollments for course_id 2 also deleted.")
```

### 5. SQL Queries: SELECT, WHERE, ORDER BY, GROUP BY, COUNT, AVG

```python
# SELECT: Display all students
print("All Students:")
display(pd.read_sql_query("SELECT * FROM Students", conn))

# SELECT with WHERE: Students older than 20
print("Students older than 20:")
display(pd.read_sql_query("SELECT name, age FROM Students WHERE age > 20", conn))

# SELECT with ORDER BY: Courses ordered by name descending
print("Courses ordered by name (descending):")
display(pd.read_sql_query("SELECT course_name, instructor FROM Courses ORDER BY course_name DESC", conn))

# SELECT with GROUP BY and COUNT: Number of courses per instructor
print("Number of courses per instructor:")
display(pd.read_sql_query("SELECT instructor, COUNT(course_id) AS num_courses FROM Courses GROUP BY instructor", conn))

# SELECT with AVG: Average age of students
print("Average age of students:")
display(pd.read_sql_query("SELECT AVG(age) AS average_student_age FROM Students", conn))
```

### 6. Perform an INNER JOIN between Students, Enrollments, and Courses

```python
# INNER JOIN: Display students and the courses they are enrolled in
print("Students and their enrolled courses:")
display(pd.read_sql_query(
    """SELECT
        S.name AS StudentName,
        S.age AS StudentAge,
        C.course_name AS CourseName,
        C.instructor AS Instructor
    FROM
        Students AS S
    INNER JOIN
        Enrollments AS E ON S.student_id = E.student_id
    INNER JOIN
        Courses AS C ON E.course_id = C.course_id
    ORDER BY
        S.name, C.course_name
    """, conn
))

# Close the connection (important for saving if not in-memory, but good practice)
conn.close()
print("Database connection closed.")
```
